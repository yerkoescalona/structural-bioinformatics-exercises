# Candidate proteins for your capstone project

Browse these and pick the one you want to work with. It stays with you through the whole
course.

**You choose the protein. The ligand comes with it and is not a choice.** Each entry names
one ligand, and that is the one this course has already run end to end with that protein:
molecular dynamics, docking, and a simulation of the two together. It works.

That matters more than it looks. Most PDB entries also contain buffer components, ions,
cryoprotectants and crystallisation additives, and in a viewer they look exactly like a
substrate. If you pick a different het group from the same crystal, the later exercises will
fail, or worse, they will run and give you a number that means nothing. Use the ligand code
given here.

Everything else about your structure, the method it was solved by, its resolution, how many
residues it has, how confident AlphaFold is about it, you work out yourself. That is the
exercise.

Nothing needs to be run to read this page. The optional cell at the end opens any structure
in a 3D viewer.
